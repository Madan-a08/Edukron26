```python
# Question 1
# numbers = [10, 20, 30, 40, 50]
# Using reduce(), calculate the sum of all numbers.
from functools import reduce
number =[10,20,30,40,50]
result= reduce(lambda x,y : x+y, number)
print(result)
```

```python
# Question 2
# numbers = [2, 3, 4, 5]
# Using reduce(), calculate the product of all numbers.

numbers =[2,3,4,5]
result =reduce(lambda x,y: x*y, numbers)
print(result)
```

```python
# Question 3
# numbers = [45, 12, 89, 34, 67, 91, 28]
# Using reduce(), find the maximum number.

numbers=[45,12,89,34,67,91,28]
result = reduce(lambda x,y: x if x > y else y, numbers)
print(result)
```

```python
# Question 4
# numbers = [45, 12, 89, 34, 67, 91, 28]
# Using reduce(), find the minimum number.

numbers=[45,12,89,34,67,91,28]
result =reduce(lambda x,y: x if x < y else y,  numbers)
print(result)
```

```python
# Question 5
# words = ["Python", " ", "is", " ", "powerful"]
# Using reduce(), combine all strings into one sentence.

words= ["Python", " ", "is", " ", "powerful"]
result = reduce(lambda x, y: x+y, words)
print(result)
```

```python
# Question 6
# words = ["Python", "Azure", "Cloud", "AI"]
# Using reduce(), calculate the total number of characters in all words.

words= ["Python", "Azure", "Cloud", "AI"]
result= reduce(lambda x, y: x + len(y),words[1:], len(words[0]))
print(result)
```

```python
# Question 7
# number = 6
# numbers = list(range(1, number + 1))
# Using reduce(), calculate the factorial of number.

number=6
numbers = list(range(1, number+1))
result = reduce(lambda x, y: x*y, numbers)
print(result)
```

```python
# Question 1
# numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
# Using filter() and map():

# Extract all even numbers.
# Calculate the square of every even number.
# Return the squared values as a list.

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
result=list(map(lambda x: x**2, filter(lambda x: x%2==0, numbers)))
print(result)
```

```python
# Question 2
# numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
# Using filter() and reduce():

# Extract all odd numbers.
# Calculate their total sum.

numbers =[12,45,67,89,23,56,91,34,78,100]
result=reduce(lambda x, y: x + y, filter(lambda x:x%2 != 0, numbers))
print(result)
```
