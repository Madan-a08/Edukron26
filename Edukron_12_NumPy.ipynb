NumPy Retail Analytics Assignment
Business Scenario
You are working as a Data Analyst for SmartMart Retail Stores. The company operates five stores in different cities and sells six major product categories.

Management wants to analyze sales performance, expenses, profits, targets, discounts, inventory, and store-level performance using NumPy.

You are provided with two main NumPy arrays:

sales_data — monthly sales revenue generated by each store for different product categories.
cost_data — total cost incurred by each store for the same product categories.

Part 1: Dataset Creation

```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

Part 2: Basic Array Understanding

```python
# Question 1: Array Dimensions
# The finance manager wants to confirm how many stores and product categories are present in the data.

# Write NumPy code to display:

# Number of dimensions in sales_data
# Shape of sales_data
# Total number of values
# Data type of the array
# Memory occupied by the array
print("dimension:", np.ndim(sales_data))
print("shape: " ,np.shape(sales_data))
print("values: ",np.size(sales_data))
print("data type: ",sales_data.dtype)
print("memory: ",sales_data.itemsize)

```

```python
# Question 2: Store and Category Count
# Calculate the following:

# Total number of stores
# Total number of product categories
# Total number of store-category combinations

print("total number of stores: ", np.size(stores))
print("total number of product categories: ",np.size(categories))
print("total number of store-category combinations: ",np.size(stores)*np.size(categories))
```

```python
# Question 3: Sales Data Type Conversion
# The reporting application requires sales values in floating-point format.

# Convert sales_data into:

# float32
# float64
# Compare the memory used by both arrays.
sales_data_float32 = np.array(sales_data, dtype=np.float32)
sales_data_float64 = np.array(sales_data, dtype=np.float64)
print(sales_data_float32.dtype)
print(sales_data_float64.dtype)
print(sales_data_float32.itemsize)
print(sales_data_float64.itemsize)
print("memory float32: ",sales_data_float32.nbytes)
print("memory float64:",sales_data_float64.nbytes)
```

```python
# Part 3: Indexing and Data Extraction
# Question 4: Bengaluru Store Analysis
# The Bengaluru regional manager wants to review sales for all product categories.
# Extract and display the complete sales row for Bengaluru.
print(sales_data[0])
```

```python
# Question 5: Mumbai Cost Analysis
# Extract the total cost incurred by the Mumbai store for all product categories.
print(cost_data[3].sum())
```

```python
# Question 8: Hyderabad Furniture Performance
# Find the following for Hyderabad:

# Furniture sales
# Furniture cost
# Furniture profit

print("furniture sales: ",sales_data[1:,3])
print("furniture cost: ",cost_data[1:,3])
print("furniture profit: ", sales_data[1:,3] - cost_data[1:,3])
```

```python
# Question 9: Chennai Category Range
# Extract Chennai’s sales for the following categories:

# Groceries
# Clothing
# Furniture
# Use array slicing.
print(sales_data[2, 1:4])
```

```python
# Question 10: Selected Store Analysis
# Extract sales data for:
# Bengaluru
# Chennai
# Pune
# Include only:
# Electronics
# Clothing
# Beauty
# Use advanced indexing.

print(sales_data[[0,2,4]][:,[0,2,4]])
```

```python
# Part 4: Sales Aggregation
# Question 11: Company-Wide Sales
# Calculate the total sales revenue generated by all stores and all categories.
print(sales_data.sum())
```

```python
# Question 12: Store-Wise Sales
# Calculate total sales revenue for each store.
# Display the result together with the store names.
print("bengalore :",sales_data[0].sum())
print("hyderabad :", sales_data[1].sum())
print("chennai :", sales_data[2].sum())
print("mumbai :",sales_data[3].sum())
print("pune :" , sales_data[4].sum())
```

```python
# Question 13: Category-Wise Sales
# Calculate total sales for each product category across all stores.

# Display each category name with its total sales.
print("electronics :", sales_data[:, 0].sum())
print("groceries :", sales_data[:, 1].sum())
print("clothing :", sales_data[:,2].sum())
print("furniture :",sales_data[:,3].sum())
print("beauty :", sales_data[:,4].sum())
print("sports :",sales_data[:,5].sum())
```

```python
# Question 14: Average Sales Per Store
# Calculate the average sales generated by each store across all six product categories.
print("bengalore :",sales_data[0].sum()/sales_data.shape[1])
print("hyderabad :", sales_data[1].sum()/sales_data.shape[1])
print("chennai :", sales_data[2].sum()/sales_data.shape[1])
print("mumbai :",sales_data[3].sum()/sales_data.shape[1])
print("pune :" , sales_data[4].sum()/sales_data.shape[1])
```

```python
# Question 15: Average Sales Per Category
# Calculate the average sales of each product category across the five stores.
print("electronics :", sales_data[:, 0].sum()/sales_data.shape[0])
print("groceries :", sales_data[:, 1].sum()/sales_data.shape[0])
print("clothing :", sales_data[:,2].sum()/sales_data.shape[0])
print("furniture :",sales_data[:,3].sum()/sales_data.shape[0])
print("beauty :", sales_data[:,4].sum()/sales_data.shape[0])
print("sports :",sales_data[:,5].sum()/sales_data.shape[0])
```

```python
# Question 16: Highest-Selling Store
# Identify:

# Index of the store with the highest total sales
# Name of the store
# Total sales generated by that store
# Use np.argmax().

value = np.argmax(sales_data.sum(axis=1))
print(value)
print(stores[value])
print(sales_data[value].sum())
```

```python
# Question 17: Lowest-Selling Store
# Identify the store with the lowest overall sales.

# Use np.argmin().
value = np.argmin(sales_data.sum(axis=1))
print(value)
print(stores[value])
print(sales_data[value].sum())
```

```python
# Question 18: Highest-Selling Category
# Find the product category that generated the highest total sales across the company.
index =np.argmax(sales_data.sum(axis =0))
print(index)
print(categories[index])
print(sales_data[:,index].sum())
```

```python
# Question 19: Lowest-Selling Category
# Find the product category with the lowest total sales across all stores
index = np.argmin(sales_data.sum(axis =0))
print(index)
print(categories[index])
print(sales_data[:,index].sum())
```

```python
# Part 5: Cost Analysis
# Question 20: Total Company Cost
# Calculate the total cost incurred by SmartMart across all stores and product categories.
print(cost_data.sum())
```

```python
# Question 21: Store-Wise Cost
# Calculate the total cost incurred by each store.
print("bengalore :" ,cost_data[0].sum())
print("hyderbad :", cost_data[1].sum())
print("chennai :", cost_data[2].sum())
print("mumbai :",cost_data[3].sum())
print("pune :",cost_data[4].sum())
```

```python
# Question 22: Category-Wise Cost
# Calculate the total cost incurred for each product category across all stores.

print("electronics :", cost_data[:, 0].sum())
print("groceries :", cost_data[:, 1].sum())
print("clothing :", cost_data[:,2].sum())
print("furniture :",cost_data[:,3].sum())
print("beauty :", cost_data[:,4].sum())
print("sports :",cost_data[:,5].sum())
```

```python
# Question 23: Most Expensive Store
# Identify the store with the highest total operating cost.

Expensive =np.argmax(cost_data.sum(axis=1))
print(stores[Expensive])
```

```python
# Question 24: Lowest-Cost Store
# Identify the store with the lowest total cost.
min = np.argmin(cost_data.sum(axis=1))
print(stores[min])
```

```python
# Question 25: Most Expensive Product Category
# Determine which product category has the highest total cost across all stores.
index =np.argmax(cost_data.sum(axis=0))
print(categories[index])
```

```python
# Part 6: Profit Analysis Using Two Arrays
# Question 26: Profit Matrix
# Calculate the profit generated for every store and product category.

# Use:

# profit_data = sales_data - cost_data
# Display the complete profit matrix.

profit_data = sales_data - cost_data
print(profit_data)
```

```python
# Question 27: Store-Wise Profit
# Calculate the total profit generated by each store.
print("bengalore :",profit_data[0].sum())
print("hyderabad :", profit_data[1].sum())
print("chennai :", profit_data[2].sum())
print("mumbai :",profit_data[3].sum())
print("pune :",profit_data[4].sum())
```

```python
# Question 28: Category-Wise Profit
# Calculate the total profit generated by each product category.
print("electronics :", profit_data[:, 0].sum())
print("groceries :", profit_data[:, 1].sum())
print("clothing :", profit_data[:,2].sum())
print("furniture :",profit_data[:,3].sum())
print("beauty :", profit_data[:,4].sum())
print("sports :",profit_data[:,5].sum())
```

```python
# Question 29: Most Profitable Store
# Identify the store that generated the highest total profit.

# Display:

# Store name
# Total sales
# Total cost
# Total profit

index=np.argmax(profit_data.sum(axis=1))
print("store;", stores[index])
print("total sales:",sales_data[index].sum())
print(cost_data[index].sum())
print(sales_data[index].sum()-cost_data[index].sum())
```

```python
# Question 30: Least Profitable Store
# Identify the store that generated the lowest total profit.

least = np.argmin(profit_data.sum(axis=1))
print("store: ", stores[least])
```

```python
# Question 31: Most Profitable Category
# Find the category with the highest total profit across all stores.
index = np.argmax(profit_data.sum(axis=0))
print("category: ",categories[index])
```

```python
# Question 32: Least Profitable Category
# Find the category with the lowest total profit.

lesat=np.argmin(profit_data.sum(axis=0))
print("category: ",categories[lesat])
```

```python
# Question 33: Maximum Individual Profit
# Find the highest profit generated by any single store-category combination.

# Display:

# Profit amount
# Store name
# Product category
# Use np.unravel_index().
index= np.argmax(profit_data)
r , c = np.unravel_index(index, profit_data.shape)
print("profit:",profit_data[r,c])
print("store:",stores[r])
print("category:", categories[c])
```

```python
# Question 34: Minimum Individual Profit
# Find the smallest profit generated by a store-category combination.

# Display the corresponding store and category.
index= np.argmin(profit_data)
r,c = np.unravel_index(index, profit_data.shape)
print("profit:",profit_data[r,c])
print("store:",stores[r])
print("category:", categories[c])
```

```python
# Part 7: Profit Margin Analysis
# Question 35: Profit Margin Matrix
# Calculate the profit margin percentage for every store-category combination.

# Use:

# profit_margin = (profit_data / sales_data) * 100
# Round the result to two decimal places.
profit_margin = (profit_data / sales_data) * 100
print(profit_margin.round(2))
```

```python
# Question 36: Store-Wise Profit Margin
# Calculate the overall profit margin percentage for each store using:

# Store Profit Margin =
# Total Store Profit / Total Store Sales × 100
store_profit_margin = (profit_data.sum(axis=1)/sales_data.sum(axis=1))*100
print(store_profit_margin.round(2))
```

```python
# Question 37: Category-Wise Profit Margin
# Calculate the overall profit margin percentage for each product category.

category_profit_margin=(profit_data.sum(axis=0)/sales_data.sum(axis=0))*100
print(category_profit_margin.round(2))
```

```python
# Question 38: Best Store by Profit Margin
# Identify the store with the highest overall profit margin.
index = np.argmax(store_profit_margin)
print("store:",stores[index])
```

```python
# Question 39: Best Category by Profit Margin
# Identify the product category with the highest profit margin.
index=np.argmax(category_profit_margin)
print("category:",categories[index])
```

```python
# Question 40: Low-Margin Transactions
# Management considers a profit margin below 22% to be weak.

# Find all store-category combinations where the profit margin is below 22%.

# Display:

# Store
# Category
# Sales
# Cost
# Profit
# Profit margin
# Use Boolean indexing.
low = profit_margin < 22
r,c = np.where(low)
for r,c in zip(r,c):
  print("store:",stores[r])
  print("category:",categories[c])
  print("sales:",sales_data[r,c])
  print("cost:",cost_data[r,c])
  print("profit:",profit_data[r,c])
  print("profit margin:",profit_margin[r,c])
```

```python
# Part 8: Boolean Filtering
# Question 41: High Sales Values
# Find all sales values greater than ₹5,00,000.

high=sales_data[sales_data> 500000]
print(high)
```

```python
# Question 42: Low Sales Values
# Find all sales values below ₹2,00,000.
low = sales_data[sales_data<200000]
print(low)
```

```python
# Question 43: High Profit Values
# Find all store-category combinations where profit is greater than ₹1,50,000.

high =profit_data[profit_data>150000]
print(high)
```

```python
# Question 44: Moderate Sales Range
# Find sales values between ₹2,50,000 and ₹5,00,000.
# Use np.logical_and().

moderate=sales_data[np.logical_and(sales_data>= 250000,sales_data<=500000) ]
print(moderate)
```

```python
# Question 45: Premium Performance
# A store-category combination is considered premium when:

# Sales are greater than ₹5,00,000
# Profit is greater than ₹1,50,000
# Find all premium-performing combinations.
premium=profit_data[np.logical_and(sales_data>500000,profit_data>150000)]
print(premium)
```
