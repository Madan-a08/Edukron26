NumPy Retail Analytics Assignment
Business Scenario
You are working as a Data Analyst for SmartMart Retail Stores. The company operates five stores in different cities and sells six major product categories.

Management wants to analyze sales performance, expenses, profits, targets, discounts, inventory, and store-level performance using NumPy.

You are provided with two main NumPy arrays:

sales_data — monthly sales revenue generated by each store for different product categories.
cost_data — total cost incurred by each store for the same product categories.

Part 1: Dataset Creation

```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

```python
# Question 27: Store-Wise Profit
# Calculate the total profit generated by each store.
print("bengalore :",profit_data[0].sum())
print("hyderabad :", profit_data[1].sum())
print("chennai :", profit_data[2].sum())
print("mumbai :",profit_data[3].sum())
print("pune :",profit_data[4].sum())
```

```python
# Question 28: Category-Wise Profit
# Calculate the total profit generated by each product category.
print("electronics :", profit_data[:, 0].sum())
print("groceries :", profit_data[:, 1].sum())
print("clothing :", profit_data[:,2].sum())
print("furniture :",profit_data[:,3].sum())
print("beauty :", profit_data[:,4].sum())
print("sports :",profit_data[:,5].sum())
```

```python
# Question 29: Most Profitable Store
# Identify the store that generated the highest total profit.

# Display:

# Store name
# Total sales
# Total cost
# Total profit

index=np.argmax(profit_data.sum(axis=1))
print("store;", stores[index])
print("total sales:",sales_data[index].sum())
print(cost_data[index].sum())
print(sales_data[index].sum()-cost_data[index].sum())
```

```python
# Question 30: Least Profitable Store
# Identify the store that generated the lowest total profit.

least = np.argmin(profit_data.sum(axis=1))
print("store: ", stores[least])
```

```python
# Question 32: Least Profitable Category
# Find the category with the lowest total profit.

lesat=np.argmin(profit_data.sum(axis=0))
print("category: ",categories[lesat])
```

```python
# Question 33: Maximum Individual Profit
# Find the highest profit generated by any single store-category combination.

# Display:

# Profit amount
# Store name
# Product category
# Use np.unravel_index().
index= np.argmax(profit_data)
r , c = np.unravel_index(index, profit_data.shape)
print("profit:",profit_data[r,c])
print("store:",stores[r])
print("category:", categories[c])
```

```python
# Question 34: Minimum Individual Profit
# Find the smallest profit generated by a store-category combination.

# Display the corresponding store and category.
index= np.argmin(profit_data)
r,c = np.unravel_index(index, profit_data.shape)
print("profit:",profit_data[r,c])
print("store:",stores[r])
print("category:", categories[c])
```

```python
# Part 7: Profit Margin Analysis
# Question 35: Profit Margin Matrix
# Calculate the profit margin percentage for every store-category combination.

# Use:

# profit_margin = (profit_data / sales_data) * 100
# Round the result to two decimal places.
profit_margin = (profit_data / sales_data) * 100
print(profit_margin.round(2))
```

```python
# Question 36: Store-Wise Profit Margin
# Calculate the overall profit margin percentage for each store using:

# Store Profit Margin =
# Total Store Profit / Total Store Sales × 100
store_profit_margin = (profit_data.sum(axis=1)/sales_data.sum(axis=1))*100
print(store_profit_margin.round(2))
```

```python
# Question 37: Category-Wise Profit Margin
# Calculate the overall profit margin percentage for each product category.

category_profit_margin=(profit_data.sum(axis=0)/sales_data.sum(axis=0))*100
print(category_profit_margin.round(2))
```

```python
# Question 38: Best Store by Profit Margin
# Identify the store with the highest overall profit margin.
index = np.argmax(store_profit_margin)
print("store:",stores[index])
```

```python
# Question 39: Best Category by Profit Margin
# Identify the product category with the highest profit margin.
index=np.argmax(category_profit_margin)
print("category:",categories[index])
```

```python
# Question 40: Low-Margin Transactions
# Management considers a profit margin below 22% to be weak.

# Find all store-category combinations where the profit margin is below 22%.

# Display:

# Store
# Category
# Sales
# Cost
# Profit
# Profit margin
# Use Boolean indexing.
low = profit_margin < 22
r,c = np.where(low)
for r,c in zip(r,c):
  print("store:",stores[r])
  print("category:",categories[c])
  print("sales:",sales_data[r,c])
  print("cost:",cost_data[r,c])
  print("profit:",profit_data[r,c])
  print("profit margin:",profit_margin[r,c])
```

```python
# Question 44: Moderate Sales Range
# Find sales values between ₹2,50,000 and ₹5,00,000.
# Use np.logical_and().

moderate=sales_data[np.logical_and(sales_data>= 250000,sales_data<=500000) ]
print(moderate)
```

```python
# Question 45: Premium Performance
# A store-category combination is considered premium when:

# Sales are greater than ₹5,00,000
# Profit is greater than ₹1,50,000
# Find all premium-performing combinations.
premium=profit_data[np.logical_and(sales_data>500000,profit_data>150000)]
print(premium)
```

```python
# Question 46: Underperforming Combinations
# A store-category combination is considered underperforming when:
# Sales are below ₹2,50,000
# Profit margin is below 25%
# Find all such combinations.

under = profit_margin[np.logical_and(sales_data<250000,profit_margin<25)]
print(under)
```

```python
# Question 47: Category Performance Classification
# Create a new array that classifies every sales value as:

# "High" when sales are at least ₹5,00,000
# "Medium" when sales are between ₹2,50,000 and ₹4,99,999
# "Low" when sales are below ₹2,50,000
# Use np.where() or np.select().

conditions= [sales_data>= 500000,sales_data>=250000]
c = ["high", "medium"]
classi = np.select(conditions, c, default="low")
print(classi)
```

```python
# Part 9: Target-Based Analysis
# Use the following category-wise monthly sales targets:

sales_targets = np.array([
    800000,  # Electronics
    450000,  # Groceries
    320000,  # Clothing
    300000,  # Furniture
    200000,  # Beauty
    230000   # Sports
])
```

```python
# Question 48: Target Comparison
# Compare every store’s category sales against sales_targets.

# Use NumPy broadcasting.
target = sales_data - sales_targets
print(target)
```

```python
# Question 49: Target Achievement Matrix
# Create a Boolean array where:

# True means the store achieved the category target
# False means the store failed to achieve the target
target_martix = sales_data >= sales_targets
print(target_martix)
```

```python
# Question 50: Target Achievement Count
# Calculate the number of categories in which each store achieved its target.

target_count =np.sum(target_martix, axis=1)
print(target_count)
```

```python
# Question 51: Best Target-Performing Store
# Find the store that achieved the highest number of category targets.

index=np.argmax(target_count)
print(stores[index])
```

```python
# Question 52: Target Achievement by Category
# Calculate how many stores achieved the target for each product category.
target_count =np.sum(target_martix, axis=0)
print(target_count)
```

```python
# Question 53: Sales Target Variance
# Calculate the difference between actual sales and target sales for every store.
# Use:
# Variance = Actual Sales - Target Sales
variance =sales_data-sales_targets
print(variance)
```

```python
# Question 54: Percentage Above or Below Target
# Calculate the percentage variance from the target using:

# Percentage Variance =
# (Actual Sales - Target Sales) / Target Sales × 100

percentage = (sales_data-sales_targets)/sales_targets*100
print(percentage)
```

```python
# Question 55: Stores Missing Electronics Target
# Identify stores that failed to achieve the Electronics sales target.
electronics_target = sales_targets[0]
result = stores[sales_data[:, 0] < electronics_target]
print(result)
```

```python
# Question 56: Stores Achieving All Targets
# Check whether any store achieved its target in all six product categories.
# Use np.all().

all= np.all(target_martix, axis=1)
print(stores[all])
```

```python
# Question 57: Stores Achieving At Least One Target
# Identify stores that achieved at least one category target.
# Use np.any().
x = np.any(target_martix, axis=1)
print(stores[x])
```

```python
# Part 10: Discount Scenario
# SmartMart plans to offer category-wise discounts:

discount_rates = np.array([
    0.10,  # Electronics: 10%
    0.05,  # Groceries: 5%
    0.15,  # Clothing: 15%
    0.12,  # Furniture: 12%
    0.08,  # Beauty: 8%
    0.10   # Sports: 10%
])
```

```python
# Question 58: Discount Amount Matrix
# Calculate the discount amount for each store-category combination.
# Use broadcasting.
dis=sales_data*discount_rates
print(dis)
```

```python
# Question 59: Sales After Discount
# Calculate the net sales value after applying the category-wise discounts.
after_dis = sales_data - dis
print(after_dis)
```

```python
# Question 60: Profit After Discount
#  Assuming costs remain unchanged, calculate the new profit after the discounts are applied.
profit =after_dis-cost_data
print(profit)
```

```python
# Question 61: Profit Margin After Discount
# Calculate the new profit margin after applying discounts.
profit_margin=(profit/after_dis)*100
print(profit_margin)
```

```python
# Question 62: Loss-Making Transactions
# Check whether any store-category combination becomes loss-making after discounts.

# Display the affected store and category names.
loss= profit<0
row, col=np.where(loss)
for row, col in zip(row, col):
  print("store:",stores[row])
  print("category:",categories[col])
```

```python
# Question 63: Discount Impact
# Calculate how much total revenue the company loses because of the proposed discounts.
print(dis.sum())
```

```python
# Question 64: Store Most Affected by Discounts
# Identify the store that loses the highest amount of revenue due to discounts.

print(stores[np.argmax(dis.sum(axis=1))])
```

```python
# Part 11: Tax and Price Adjustment
# The government introduces category-wise tax rates:

tax_rates = np.array([
    0.18,
    0.05,
    0.12,
    0.18,
    0.18,
    0.12
])
```

```python
# Question 65: Tax Collection Matrix
# Calculate the tax collected on sales for each store and category.
tax = sales_data*tax_rates
print(tax)
```

```python
# Question 66: Sales Including Tax
# Calculate the final customer billing value after adding tax.
sales= sales_data+tax
print(sales)
```

```python
# Question 67: Store-Wise Tax Collection
# Calculate the total tax collected by each store.
tax_store=tax.sum(axis=1)
print(tax_store)
```

```python
# Question 68: Category-Wise Tax Collection
# Calculate total tax collected for each category.
tax_category= tax.sum(axis=0)
print(tax_category)
```

```python
# Question 69: Highest Tax-Contributing Store
# Identify the store that contributes the highest tax amount.
print(stores[np.argmax(tax_store)])
```
