```python
class Account:
    def __init__(self, account_number, balance=0):
        self.account_number=account_number
        self.__balance=balance

    @staticmethod
    def validate_account_reference(account_number):
        if not isinstance(account_number,str):
            return False
        if account_number=="":
            return False
        if not account_number.startswith("ACC-"):
            return False
        if not account_number[4:].isdigit():
            return False
        return True

    def show_balance(self):
        print("Account:",self.account_number)
        print("Balance:",self.__balance)

account_number="ACC-1011"

if Account.validate_account_reference(account_number):
    account=Account(account_number,5000)
    account.show_balance()
else:
    print("Invalid account reference")
```

```python
class Account:
    def __init__(self, account_number, balance):
        if not account_number:
            raise ValueError("Account reference cannot be empty")

        if not isinstance(balance,(int,float)):
            raise TypeError("Balance must be numeric")

        if balance<0:
            raise ValueError("Balance cannot be negative")

        self.account_number=account_number
        self.__balance=balance

    def withdraw(self, amount):
        if not isinstance(amount,(int,float)):
            raise TypeError("Amount must be numeric")

        if amount<=0:
            raise ValueError("Amount must be greater than zero")

        if amount>self.__balance:
            raise ValueError("Insufficient balance")

        self.__balance=self.__balance-amount

    def deposit(self, amount):
        if not isinstance(amount,(int,float)):
            raise TypeError("Amount must be numeric")

        if amount<=0:
            raise ValueError("Amount must be greater than zero")

        self.__balance=self.__balance+amount

    def get_balance(self):
        return self.__balance


class UnitOfWork:
    def __init__(self, accounts):
        self.accounts=accounts
        self.old_balances={account.account_number:account.get_balance() for account in accounts}

    def __enter__(self):
        print("Transaction started")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            for account in self.accounts:
                account._Account__balance=self.old_balances[account.account_number]

            print("Transaction rolled back")
            print("Reason:",exc_value)
            return True

        print("Transaction committed")


try:
    account1=Account("ACC-1075",5000)
    account2=Account("ACC-2075",3000)

    with UnitOfWork([account1,account2]):
        account1.withdraw(2000)
        account2.deposit(2000)

    print("Account:",account1.account_number)
    print("Final balance:",account1.get_balance())
    print("Account:",account2.account_number)
    print("Final balance:",account2.get_balance())

except (TypeError,ValueError) as e:
    print("Transaction failed:",e)
```
