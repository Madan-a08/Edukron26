```python
# Assignment 12: Check Leap Year
# Write a function named is_leap_year().

# Requirements
# Accept a year as an integer.
# Return True when it is a leap year.
# Otherwise, return False.
# Print an appropriate message.

def is_leap_year(year):
  if year%4 == 0 or year%400 == 0 and year%100 != 0:
    return True
  else:
    return False
year = int(input("enter the year: "))
if is_leap_year(year):
  print(year,"is a leap year")
else:
  print(year, "is not a leap year")
```

```python
# Assignment 13: Factorial Using a Function
# Write a function named calculate_factorial().

# Requirements
# Accept a non-negative integer.
# Calculate its factorial using a loop.
# Do not use math.factorial().
# Return the factorial.
# Handle negative input.

def calculate_factorial(n):
  if n<0:
    return "Please enter a non-negative integer"
  elif n == 0:
    return 1
  else:
    return n*calculate_factorial(n-1)
n = int(input("enter the number: "))
print("Factorial: ", calculate_factorial(n))
```

```python
# Assignment 15: Count Vowels in a String
# Write a function named count_vowels().

# Requirements
# Accept a string.
# Count uppercase and lowercase vowels.
# Return the total vowel count.
# Do not use separate count statements for every vowel.

def count_vowels(s):
  count = 0
  for i in s:
    if i in "aeiouAEIOU":
      count = count + 1
  return count
s = input("enter the string: ")
print("Total count: ",count_vowels(s))
```

```python
# Assignment 17: Palindrome Checker
# Write a function named is_palindrome().

# Requirements
# Accept a string.
# Ignore spaces and letter case.
# Return True when the string is a palindrome.
# Print a meaningful result.

def is_palindrome(s):
  s=s.replace(" ","").lower()
  return s == s[::-1]
s = input("enter the string: ")
if is_palindrome(s):
  print(s, "is a palindrome")
else:
  print(s, "is not a palindrome")
```

```python
# Assignment 19: Find Maximum in a List
# Write a function named find_list_maximum().

# Requirements
# Accept a list of numbers.
# Do not use max().
# Return the largest list element.
# Handle an empty list by returning an appropriate message.

def find_list_maximum(l):
  if len(l)==0:
    return "Empty list"
  else:
    max = l[0]
    for i in l:
      if i > max:
        max = i
    return max
lst = list(map(int, input().split()))
print("maximum value: ", find_list_maximum(lst))
```

```python
# Assignment 20: Count Positive, Negative and Zero Values
# Write a function named count_number_types().

# Requirements
# Accept a list of integers.
# Count positive numbers, negative numbers, and zeros.
# Return all three counts.
# Use tuple unpacking to receive the returned values.

def count_number_types(l):
  pos = 0
  neg = 0
  zeros = 0
  for i in l:
    if i>0:
      pos = pos +1
    elif i<0:
      neg = neg+1
    else :
      zeros = zeros + 1
  return pos, neg, zeros
l = list(map(int, input().split()))
positive, negative, zeros = count_number_types(l)
print("Positive: ", positive)
print("Negative: ", negative)
print("Zeros: ", zeros)
```

```python
# Assignment 24: Variable-Length Positional Arguments
# Write a function named calculate_total() using *args.

# Requirements
# Accept any number of numerical values.
# Return their total.
# Do not use sum().
# Call the function with 3, 5, and 10 arguments.

def calculate_total(*args):
  total= 0
  for i in args:
    total = total + i
  return total

print(calculate_total(6, 8,23))
print(calculate_total(45,3,67,43,23))
print(calculate_total(2,4,5,7,43,55,32,23,45,49))
```

```python
# Assignment 25: Variable-Length Keyword Arguments
# Write a function named display_student() using **kwargs.

# Requirements
# Accept any number of student attributes.
# Print every key and value.
# Call it with name, age, course, city, and qualification.
# Call it again using a different set of fields.

def display_student(**kwargs):
  for key, values in kwargs.items():
    print(key, values)
  print()
display_student(name="Ram",age= 18, course = "Ai", city = "Bengaluru")
display_student(name="Rahul", age=30, course="Python", city ="Mysuru")
```
