```python
# Assignment 19: Find Maximum in a List
# Write a function named find_list_maximum().

# Requirements
# Accept a list of numbers.
# Do not use max().
# Return the largest list element.
# Handle an empty list by returning an appropriate message.

def find_list_maximum(l):
  if len(l)==0:
    return "Empty list"
  else:
    max = l[0]
    for i in l:
      if i > max:
        max = i
    return max
lst = list(map(int, input().split()))
print("maximum value: ", find_list_maximum(lst))
```

```python
# Assignment 20: Count Positive, Negative and Zero Values
# Write a function named count_number_types().

# Requirements
# Accept a list of integers.
# Count positive numbers, negative numbers, and zeros.
# Return all three counts.
# Use tuple unpacking to receive the returned values.

def count_number_types(l):
  pos = 0
  neg = 0
  zeros = 0
  for i in l:
    if i>0:
      pos = pos +1
    elif i<0:
      neg = neg+1
    else :
      zeros = zeros + 1
  return pos, neg, zeros
l = list(map(int, input().split()))
positive, negative, zeros = count_number_types(l)
print("Positive: ", positive)
print("Negative: ", negative)
print("Zeros: ", zeros)
```

```python
# Assignment 92: Closure-Based Bank Balance
# Write a function named create_bank_account().

# Requirements
# Store the balance in the outer function.
# Define inner functions for:
# Deposit
# Withdrawal
# Balance enquiry
# Use nonlocal to update the balance.
# Reject withdrawals greater than the available balance.
# Return all three inner functions.

def create_bank_account():
  balance = 0
  def deposit(amount):
    nonlocal balance
    balance = balance+ amount
  def withdraw(amount):
    nonlocal balance
    if amount > balance:
      print("insufficient balance")
    else:
      balance = balance - amount
  def check_balance():
    return balance
  return deposit, withdraw, check_balance
a,b,c = create_bank_account()
a(500)
print("balance: ", c())
b(300)
print("balance: ", c())
b(15000)
print("balance: ",c())
```

```python
# Assignment 100: Complete Functional Programming Project
# Develop a complete Employee Performance Analysis System using functions, lambda expressions, map(), filter(), reduce(), nested functions, closures, and LEGB scope.

# Sample Employee Data
# employees = [
#     {
#         "id": 101,
#         "name": "Ravi",
#         "department": "IT",
#         "salary": 60000,
#         "performance_score": 88,
#         "experience": 5
#     },
#     {
#         "id": 102,
#         "name": "Anitha",
#         "department": "HR",
#         "salary": 45000,
#         "performance_score": 72,
#         "experience": 4
#     },
#     {
#         "id": 103,
#         "name": "Kiran",
#         "department": "IT",
#         "salary": 75000,
#         "performance_score": 94,
#         "experience": 7
#     },
#     {
#         "id": 104,
#         "name": "Meena",
#         "department": "Finance",
#         "salary": 55000,
#         "performance_score": 81,
#         "experience": 6
#     },
#     {
#         "id": 105,
#         "name": "Arjun",
#         "department": "IT",
#         "salary": 40000,
#         "performance_score": 65,
#         "experience": 2
#     }
# ]
# Functional Requirements
# Part 1: Display Employees
# Write a function named display_employees() that prints all employee details in a readable format.

# Part 2: Extract Employee Names
# Use map() to extract employee names.

# Part 3: Filter High Performers
# Use filter() to select employees whose performance score is 80 or above.

# Part 4: Filter by Department
# Write a function that accepts a department name and returns matching employees using filter().

# Part 5: Salary Increase
# Use map() to calculate revised salaries based on these conditions:

# Performance score 90 or above: 15% increase
# Performance score 80–89: 10% increase
# Performance score 70–79: 5% increase
# Below 70: No increase
# Do not modify the original employee dictionaries directly.

# Part 6: Total Salary Expense
# Use reduce() to calculate:

# Current total salary
# Revised total salary
# Total additional salary expense
# Part 7: Highest-Paid Employee
# Use reduce() to find the employee with the highest salary.

# Do not use max().

# Part 8: Sort Employees
# Use sorted() with lambda expressions to sort employees by:

# Name
# Salary
# Performance score
# Experience
# Part 9: Closure-Based Bonus Calculator
# Create a function named create_bonus_calculator().

# The outer function should accept a bonus percentage.
# The inner function should accept a salary.
# Return the calculated bonus.
# Create separate bonus calculators for 5%, 10%, and 15%.
# Part 10: Nested Report Generator
# Create an outer function named generate_department_report().

# Inside it, define nested functions to calculate:

# Number of employees
# Total salary
# Average salary
# Average performance score
# Return the complete department report as a dictionary.

# Part 11: LEGB Demonstration
# Create a global variable:

# company_name = "Edukron"
# Inside the report function:

# Access the global company name.
# Create an enclosing variable for the department.
# Create a local variable for the report title.
# Add comments explaining how Python resolves each variable using LEGB.
# Part 12: Menu-Driven Program
# Create a menu containing the following options:

# 1. Display all employees
# 2. Display employee names
# 3. Display high performers
# 4. Search by department
# 5. Calculate revised salaries
# 6. Calculate total salary expense
# 7. Find highest-paid employee
# 8. Sort employees
# 9. Generate department report
# 10. Exit
from functools import reduce
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]
company_name = "Edukron"

def display_employees(emp):
  if len(emp) == 0:
    print("no employees found")
    return
  for i in emp:
    print("id:" ,i["id"],"name:",i["name"],
          "department: " ,i["department"] ,"salary: " ,i["salary"],
          "performance " ,i["performance_score"],
          "experience: " ,i["experience"])

def employee_names(emp):
  return list(map(lambda x: x["name"],  emp))

def high_performers(emp):
  return list(filter(lambda x: x["performance_score"] >= 80, emp))

def department_filter(emp, department):
  return list(filter(lambda x: x["department"] .lower()== department.lower(),emp))

def salaries(emp):
  def calculate(x):
    if x["performance_score"] >= 90:
      increase=0.15
    elif x["performance_score"] >= 80:
      increase=0.10
    elif x["performance_score"] >=70:
      increase=0.05
    else:
      increase=0
    new_salary = x["salary"] + (x["salary"] * increase)
    return {
          "name": x["name"],
        "old_salary": x["salary"],
          "new_salary": new_salary
      }
  return list(map(calculate,  emp))

def salary_expense(emp):
  current_total = reduce(lambda x, y: x + y,map(lambda x: x["salary"], emp),0)
  revised= salaries(emp)
  revised_total = reduce(lambda x, y: x + y, map(lambda x: x["new_salary"], revised), 0)
  additional_expense = revised_total - current_total
  return current_total, revised_total, additional_expense

def highest_paid(emp):
  if len(emp)==0:
    return None
  return reduce(lambda x, y: x if x["salary"] > y["salary"] else y, emp)


def sort_employees(emp):
  print("1 name , 2 salary, 3 perfeormance, 4 experience ")
  choice = input("enter sorting choice: ")
  if choice=="1":
    return sorted(emp, key=lambda x: x["name"])
  elif choice=="2":
    return sorted(emp, key=lambda x: x["salary"])
  elif choice=="3":
    return sorted(emp, key=lambda x: x["performance_score"])
  elif choice=="4":
    return sorted(emp, key=lambda x: x["experience"])
  else:
    print("invalid choice")
    return []

def create_bonus_calculator(percentage):
    def calculate_bonus(salary):
        return salary * percentage / 100
    return calculate_bonus
bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)

def generate_department_report(emp, department):
    department_name = department       # Enclosing scope
    report_title = "Department Report" # Local scope
    def employee_count():
        return len(emp)
    def total_salary():
        return reduce(
            lambda x, y: x + y,
            map(lambda x: x["salary"], emp),
            0
        )
    def average_salary():
        if len(emp) == 0:
            return 0
        return total_salary() / len(emp)
    def average_performance():
        if len(emp) == 0:
            return 0
        return reduce(
            lambda x, y: x + y,
            map(lambda x: x["performance_score"], emp))/len(emp)
    return {
        "company": company_name,
        "title": report_title,
        "department": department_name,
        "employee_count": employee_count(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_performance": average_performance()}

while True:
    print("\n1. display all employees")
    print("2. display employee names")
    print("3. display high performers")
    print("4. search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. find highest-paid employee")
    print("8. sort employees")
    print("9. generate department report")
    print("10.exit")
    choice = input("enter your choice: ")
    if choice == "1":
        display_employees(employees)
    elif choice == "2":
        print("Employee names:", employee_names(employees))
    elif choice == "3":
        display_employees(high_performers(employees))
    elif choice == "4":
        department = input("Enter department: ")
        result = department_filter(employees, department)
        if len(result) == 0:
            print("Department not found")
        else:
            display_employees(result)
    elif choice == "5":
        result = salaries(employees)
        for i in result:
            print(i)
    elif choice == "6":
        current, revised, additional = salary_expense(employees)

        print("Current total:", current)
        print("Revised total:", revised)
        print("Additional expense:", additional)
    elif choice == "7":
        result = highest_paid(employees)
        if result is None:
            print("No employees found")
        else:
            print(result)
    elif choice == "8":
        result = sort_employees(employees)
        if len(result) != 0:
            display_employees(result)
    elif choice == "9":
        department = input("Enter department: ")
        result = department_filter(employees, department)

        if len(result) == 0:
            print("Department not found")
        else:
            print(generate_department_report(result, department))

    elif choice == "10":
        print("Program ended")
        break

    else:
        print("Invalid choice")
```
