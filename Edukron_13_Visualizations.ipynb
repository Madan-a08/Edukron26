```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y, color='blue', marker='o', linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

1. The graph shows a **steady upward trend** from 10 to 40.
2. Values increase consistently as the X-axis values increase.
3. The highest value is **40** at X = 5.
4. The lowest value is **10** at X = 1.
5. No declines or fluctuations are observed throughout the trend.
6. Overall, the data indicates **continuous and stable growth**.



```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
avg_bill = tips.groupby('day', observed=False)['total_bill'].mean().sort_index()
plt.figure(figsize=(8, 4))
plt.plot(avg_bill.index, avg_bill.values, marker='o', color='royalblue', linewidth=2)
plt.title('1. Line Plot: Average Bill by Day')
plt.xlabel('Day')
plt.ylabel('Average Total Bill')
plt.grid(True, alpha=0.4)
plt.show()
```

1. The average total bill is highest on Sunday, at approximately 21.41.
2. Saturday has the second-highest average bill, at approximately 20.44.
3. Thursday has an average bill of approximately 17.68.
4. Friday has the lowest average bill, at approximately 17.15.
5. Overall, average customer spending is higher on weekends than on weekdays.


```python
avg_tip = tips.groupby('day',observed=False)['tip'].mean().sort_index()
plt.figure(figsize=(8, 4))
plt.bar(avg_tip.index, avg_tip.values, color='green', edgecolor='black')
plt.title('2. Bar Chart: Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

1. Sunday has the highest average tip, at approximately 3.26.
2. Saturday has the second-highest average tip, at approximately 2.99.
3. Thursday has an average tip of approximately 2.77.
4. Friday has the lowest average tip, at approximately 2.73.
5. Overall, customers tend to leave higher tips on weekends.


```python
sex_bill = tips.groupby('sex',observed= False)['total_bill'].mean()
plt.figure(figsize=(7, 4))
plt.bar(sex_bill.index, sex_bill.values, color=['pink', 'skyblue'])
plt.title('3. Bar Chart: Average Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Total Bill')
plt.show()
```

1. The average bill is higher for **Male customers** than for Female customers.
2. Male customers have an average bill of approximately **20.74**.
3. Female customers have an average bill of approximately **18.06**.
4. The difference between the two groups is approximately **2.68**.
5. Overall, **Male customers have higher average spending per bill** in this dataset.


```python
plt.figure(figsize=(8, 4))
plt.hist(tips['total_bill'], bins=15, color='purple', edgecolor='black')
plt.title('4. Histogram: Total Bill Distribution')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

1. Most total bills are concentrated in the lower-to-middle range of the distribution.
2. The distribution is right-skewed, with fewer customers having very high total bills.
3. The highest total bills occur less frequently than the lower bill amounts.
4. Most observations are grouped around approximately 10 to 25.
5. Overall, the histogram shows that typical customer bills are relatively moderate, with a small number of high-value bills.


```python
plt.figure(figsize=(8, 4))
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.7, color='darkorange')
plt.title('5. Scatter Plot: Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

1. Male customers have a higher average total bill than female customers.
2. The average bill for male customers is approximately 20.74.
3. The average bill for female customers is approximately 18.06.
4. The difference in average spending between the two groups is approximately 2.68.
5. Overall, male customers show higher average spending per transaction in this dataset.


```python
plt.figure(figsize=(8, 4))
for sex_group, group in tips.groupby('sex'):
    plt.scatter(group['total_bill'], group['tip'], label=sex_group, alpha=0.7)
plt.title('6. Scatter Plot: Bill vs Tip by Sex')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.legend()
plt.show()
```

1. The graph shows a positive relationship between total bill and tip.
2. As the total bill increases, the tip generally tends to increase.
3. Both male and female customers show a similar upward pattern between bill and tip.
4. Higher total bills are associated with higher tip amounts for both groups.
5. Overall, the total bill appears to be an important factor influencing the tip amount.


```python
plt.figure(figsize=(8, 4))
plt.boxplot([tips[tips['day'] == d]['total_bill'] for d in ['Thur', 'Fri', 'Sat', 'Sun']],
            labels=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('7. Box Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

1. Saturday has the widest spread in total bill values compared with the other days.
2. Sunday shows a relatively high median total bill.
3. Friday has the lowest overall median total bill among the four days.
4. Thursday has a moderate median bill with a relatively compact distribution.
5. Saturday and Sunday show greater variability in customer spending than Thursday and Friday.


```python
plt.figure(figsize=(7, 4))
plt.boxplot([tips[tips['smoker'] == s]['total_bill'] for s in ['Yes', 'No']],
            labels=['Smoker', 'Non-Smoker'])
plt.title('8. Box Plot: Total Bill by Smoker Status')
plt.ylabel('Total Bill')
plt.show()
```

1. Non-smokers have a slightly higher median total bill than smokers.
2. Smokers show a wider spread in total bill values.
3. Both groups have some high-value bills that extend beyond the main distribution.
4. The overall bill distributions for smokers and non-smokers overlap considerably.
5. Overall, smoker status does not show a very large difference in total bill amounts.


```python
plt.figure(figsize=(8, 4))
plt.violinplot([tips[tips['day'] == d]['total_bill'] for d in ['Thur', 'Fri', 'Sat', 'Sun']],
               showmeans=True)
plt.xticks([1,2,3,4], ['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('9. Violin Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

1. Smokers and non-smokers have broadly similar total bill distributions.
2. Non-smokers have a slightly higher median total bill than smokers.
3. Smokers show a wider range of total bill values.
4. Both groups contain some higher-value bills that extend beyond the main distribution.
5. Overall, smoker status does not create a major difference in total bill amounts.


```python
counts = tips['day'].value_counts().reindex(['Sun', 'Sat', 'Thur', 'Fri'])
plt.figure(figsize=(8, 4))
plt.bar(counts.index, counts.values, color='coral')
plt.title('10. Count Plot: Number of Observations by Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

1. Saturday has the highest number of observations, with 87 records.
2. Sunday has the second-highest number of observations, with 76 records.
3. Thursday has 62 observations.
4. Friday has the lowest number of observations, with only 19 records.
5. Overall, the dataset contains many more observations for Saturday and Sunday than for Friday.


```python
smoker_counts = tips['smoker'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(smoker_counts.values, labels=smoker_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('11. Pie Chart: Smoker Share')
plt.show()
```

1. Non-smokers make up the majority of the observations, at approximately 61.9%.
2. Smokers account for approximately 38.1% of the observations.
3. The number of non-smokers is significantly higher than the number of smokers.
4. There are 151 non-smokers compared with 93 smokers in the dataset.
5. Overall, the dataset contains more non-smoker observations than smoker observations.


```python
corr = tips[['total_bill', 'tip', 'size']].corr()
plt.figure(figsize=(6, 5))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.index)), corr.index)
plt.title('12. Heatmap: Correlation Matrix')
plt.show()
```

1. Total bill and tip have the strongest positive correlation, approximately 0.68.
2. Total bill and group size also show a positive correlation, approximately 0.60.
3. Tip and group size have a moderate positive correlation, approximately 0.49.
4. All three variables have positive relationships with each other.
5. Overall, larger bills tend to be associated with higher tips and larger group sizes.
