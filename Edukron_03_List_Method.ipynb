```python
# 1. Create a list of 5 employee names and add a new employee using append().
employees = ["Rahul", "Ankith", "Raju", "Rohan", "Vikram"]
employees.append("Manu")
employees
```

```python
# 2. Create a list of account numbers and add three new account numbers using extend().
accounts = [1, 2, 3]
accounts.extend([4, 5, 6])
accounts
```

```python
# 3. Insert a new customer at the beginning of a customer list using insert().
customers = ["Rahul", "Ankith", "Raju"]
customers.insert(0, "Manu")
customers
```

```python
# 4. Insert a transaction amount at index 3 in a list.
transactions = [1000, 2000, 3000, 5000]
transactions.insert(3, 4000)
transactions
```

```python
# 5. Remove the last item from a list using pop().
name = ["Rahul", "Raju", "Kiran", "Manu"]
name.pop()
name
```

```python
# 6. Remove the customer "Rahul" from a customer list using remove().
customers = ["Rahul", "Rohan"]
customers.remove("Rahul")
customers
```

```python
# 7. Count how many times the account type "Savings" appears in a list.
accounts = ["Savings", "Current", "Savings", "Loan", "Savings"]
accounts.count("Savings")
```

```python
# 8. Find the index of "Current" in a list of account types.
accounts = ["Savings", "Current", "Loan"]
accounts.index("Current")
```

```python
# 9. Reverse a list of transaction IDs using reverse().
transactions = [1001, 1002, 1003, 1004, 1005]
transactions.reverse()
transactions
```

```python
# 10. Sort a list of account balances in ascending order using sort().
balances = [850, 45000, 120000, 95000]
balances.sort()
balances
```

```python
# 11. Sort a list of customer names in descending order using sort().
customers = ["Rahul", "Rohan", "Raju"]
customers.sort(reverse=True)
customers
```

```python
# 12. Make a copy of a list using copy() and verify that modifying the copied list does not affect the original list.
customers = ["Manu", "Raju", "Rohan"]
new_customers = customers.copy()
new_customers.append("Ramu")
print(customers, new_customers)
```

```python
# 13. Clear all elements from a list using clear().
customers = ["Manu", "Raju", "Rohan"]
customers.clear()
customers
```

```python
# 14. Append a nested list containing a new customer's details to an existing customer list.
customers = [["C101", "Raju", "Savings", 10000]]
customers.append(["C102", "Manu", "Current", 120000])
customers
```

```python
# 15. Extend a list of branch names with another list of new branches.
branches = ["MG Road", "Whitefield"]
branches.extend(["Mysore", "Mandya"])
branches
```

```python
# 17. Pop the element at index 2 from a list and print the removed value.
numbers = [10, 20, 30, 40, 50]
numbers.pop(2)
```

```python
# 18. Count how many times the city "Bangalore" appears in a list.
cities = ["Bangalore", "Mysore", "Bangalore", "Delhi"]
cities.count("Bangalore")
```

```python
# 19. Reverse a list of bank names and print the result.
banks = ["SBI", "HDFC", "ICICI", "Canara"]
banks.reverse()
banks
```

```python
# 20. Create a list of loan amounts and sort it in descending order.
loans = [500000, 750000, 650000, 900000]
loans.sort(reverse=True)
loans
```
